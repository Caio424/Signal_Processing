```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Numero complexo - Forma cartesiana
a  = 3.0 # real
b  = 4.0 # imaginaria
z1 = a + b*1j
```

```python
print(type(a), type(b), type(z1))
```

```python
# Numero complexo - Forma polar
magnitude = np.sqrt(a**2 + b**2) # mag. de z1
phase     = np.arctan2(b, a) # fase de z1 (rad)
```

```python
print(magnitude, "\n", phase)
```

```python
def COMPLEXmagnitude(z, version=1):
    if version == 1:
        'Pythagoras'
        return np.sqrt(np.real(z)**2 + np.imag(z)**2)
    if version == 2:
        'abs value'
        return np.abs(z)
```

```python
print(COMPLEXmagnitude(z1, 2))
```

## Desafio
Crie funções customizadas para calcular o módulo e fase de um complexo.

Obs: ñ funções prontas!

```python
#help (np.angle)
#help (np.arctan2)
```

```python
def COMPLEXphase(z, version=2, degUnits=False):
    if version == 1:
        'arctan'
        phi = np.arctan(np.imag(z) / np.real(z))
    if version == 2:
        'arctan2'
        phi = np.arctan2(np.imag(z), np.real(z))
    if version == 3:
        'angle'
        phi = np.angle(z)
        
    if degUnits:
        'rad to deg'
        phi *= 180 / np.pi
    
    return phi
```

```python
print(COMPLEXphase(-3 -4j))
```

```python
COMPLEXphase(3 + 4j, degUnits=True)
```

# Gráfica dos números complexos

```python
# Diagrama no plano complexo
x = z1.real # real
y = z1.imag # imag

# Plot
fig, ax = plt.subplots()
ax.plot(x, y, 'ro')
ax.arrow(0, 0, x, y, color='r', linestyle='--')
ax.set_xlabel('Re{z}')
ax.set_ylabel('Im{z}')
plt.show()
```

```python
# Potencia dos numeros complexos
z2 = z1 ** 2

x1, y1 = z1.real, z1.imag

x2, y2 = z2.real, z2.imag


# Plot
fig, ax = plt.subplots()
ax.plot(x1, y1, 'ro')
ax.arrow(0, 0, x1, y1, color='r', linestyle='--', label='z1')
ax.plot(x2, y2, 'bo')
ax.arrow(0, 0, x2, y2, color='b', linestyle='--', label='z2')
ax.set_xlabel('Re{z}')
ax.set_ylabel('Im{z}')
ax.legend()
# figureFormat(ax, fig, tight=True)
plt.show()
```

```python
COMPLEXmagnitude(z2), COMPLEXphase(z2, degUnits=True)
```

```python
# Raiz dos numeros complexos
z3 = np.sqrt(z1) # raiz principal de z1

x1, y1 = z1.real, z1.imag
x3, y3 = z3.real, z3.imag

# Plot
fig, ax = plt.subplots()
ax.plot(x1, y1, 'ro')
ax.arrow(0, 0, x1, y1, color='r', linestyle='--', label='z1')
ax.plot(x3, y3, 'bo')
ax.arrow(0, 0, x3, y3, color='b', linestyle='--', label='z3')
ax.set_xlabel('Re{z}')
ax.set_ylabel('Im{z}')
ax.legend()
ax.set_xlim([-5, 5])
ax.set_ylim([-5, 5])
plt.show()
```

## Sinais periodicos e nao periodicos

```python
N = 100 # numero de amostras
n = np.arange(N)
K1 = 3
K2 = 20 # K1/K2 racional -> sinal periodico (periodo = 20)
omega = 2*np.pi * (K1 / K2) # freq. digital
phase = 0 # fase

x = np.cos(omega*n + phase)

fig, ax = plt.subplots()
ax.stem(n, x)
ax.set_xlim([0, 40])
plt.show()
```

```python
# Sinal nao periodico: omega/(2*pi) irracional (omega = 1 rad/amostra)
omega = 1
x = np.cos(omega*n + phase)

fig, ax = plt.subplots()
ax.stem(n, x)
ax.set_xlim([0, 40])
plt.show()
```
