```python
# Resposta 1
import numpy as np
import matplotlib.pyplot as plt
from IPython import display



ti = 1e-12 # tempo inicial (liberando a singularidade para x=0 em sinc(x) / x)
tf = 10  # tempo final
Na = 1000 # Numero de amostras do sinal analogico
t = np.linspace(ti, tf, Na) # vetor de tempo (analogico)

n = np.arange(0, int((tf - ti) / (1/20))) # vetor de amostras (digital)

fs = 20 # frequencia de amostragem (e.g., 5 * f)
Ts = 1. / fs # tempo de amostragem

# Plot interativo
fig, ax = plt.subplots()

for nn in n:
    sinc = np.sinc(fs * (t - nn * Ts))

    # ax.set_xlim(t[0], t[-1])
    ax.cla()
    ax.plot(t, sinc, 'b')
    
    display.display(fig)
    display.clear_output(wait = True)
    # plt.pause(0.1)
plt.show()
```

Resposta: 't' e 'sinc' possuem o mesmo tamanho (1000)

```python
# Resposta 2

import numpy as np
import matplotlib.pyplot as plt
from IPython import display



# Sinal analógico original
ti = 1e-12
tf = 10
Na = 1000
t = np.linspace(ti, tf, Na)

f = 1.5  # frequência do sinal original (em Hz)
y_original = np.sin(2 * np.pi * f * t)  # sinal senoidal

# Amostragem
fs = 20  # frequência de amostragem (> 2*f para satisfazer Nyqvist)
Ts = 1. / fs
n = np.arange(0, int((tf - ti) / Ts))
tn = n * Ts # Vetor contendo os instantes exatos de amostragem
yn = np.sin(2 * np.pi * f * tn) # Valores matematicamente exatos

# Reconstrução usando trem de sincs
y_reconstructed = np.zeros_like(t)

fig, ax = plt.subplots()

for nn in n:
    sinc = yn[nn] * np.sinc((t - nn*Ts) / Ts)  # sinc ponderado pela amostra
    y_reconstructed += sinc
    
    # Plot animado
    ax.cla()
    ax.plot(t, y_original, 'g--', label='Original', alpha=0.7)
    ax.plot(t, y_reconstructed, 'b', label='Reconstruído', linewidth=2)
    ax.plot(nn*Ts, yn[nn], 'ro', markersize=8, label='Amostras')
    ax.set_xlim(t[0], t[-1])
    ax.legend()
    
    display.display(fig)
    display.clear_output(wait = True)

plt.show()

# Plot final comparando original vs reconstruído
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 8))

ax1.plot(t, y_original, 'g', linewidth=2, label='Sinal Original')
ax1.plot(n*Ts, yn, 'ro', markersize=6, label='Amostras')
ax1.set_title('Sinal Original e suas Amostras')
ax1.legend()


ax2.plot(t, y_original, 'g--', linewidth=2, label='Original', alpha=0.7)
ax2.plot(t, y_reconstructed, 'b', linewidth=2, label='Reconstruído')
ax2.set_title('Comparação: Original vs Reconstruído')
ax2.legend()


plt.tight_layout()
plt.show()
```

## Desafio 2:
Encontre empiricamente a minima **frequencia de amostragem** $fs$ para o sinal que você usou no exemplo anterior

```python
# Resposta

import numpy as np
import matplotlib.pyplot as plt

# Mesmo sinal do Desafio 1
ti = 0
tf = 4
Na = 1000
t = np.linspace(ti, tf, Na)

f = 2.0  # frequência do sinal
y = np.sin(2 * np.pi * f * t)

# Testar diferentes frequências de amostragem
fs_values = [3, 4, 5, 6, 8, 10, 15, 20]
fig, axes = plt.subplots(len(fs_values), 1, figsize=(12, 12))

for idx, fs in enumerate(fs_values):
    Ts = 1. / fs
    Nd = int(fs / f)
    yn = y[::Nd]
    n = np.arange(len(yn))
    
    # Reconstruir
    x_reconstructed = np.zeros_like(t)
    for nn in n:
        sinc = yn[nn] * np.sinc((t - nn*Ts) / Ts)
        x_reconstructed += sinc
    
    # Plot
    ax = axes[idx]
    ax.plot(t, y, 'g--', linewidth=2, label='Original', alpha=0.7)
    ax.plot(t, x_reconstructed, 'b', linewidth=2, label='Reconstruído')
    ax.plot(n*Ts, yn, 'ro', markersize=6)
    
    # Calcular erro
    erro = np.mean((y - x_reconstructed)**2)
    
    # Verificar se satisfaz Nyqvist (fs >= 2*f)
    nyqvist_ok = "✓" if fs >= 2*f else "✗"
    
    ax.set_title(f'fs = {fs} Hz (Nyqvist: {nyqvist_ok}) | Erro quadrático: {erro:.6f}')
    ax.legend()
    ax.set_xlim(0, 4)
    

plt.tight_layout()
plt.show()

print(f"\nSinal original tem frequência f = {f} Hz")
print(f"Frequência de Nyqvist mínima: fs_min = 2 × {f} = {2*f} Hz")
print(f"Em prática, usar fs ≥ {2*f} Hz para evitar aliasing")
```

```python
import os
import numpy as np
import matplotlib.pyplot as plt
from scipy.io import wavfile
from scipy.fft import rfft, rfftfreq
from scipy.signal import resample
from IPython.display import display, Audio


SOUND_FILE = r"C:\Users\morei\Downloads\Henrique-e-Juliano-5-KM-DVD-O-Céu-Explica-Tudo.wav"

if os.path.exists(SOUND_FILE):
    fs_original, audio_original = wavfile.read(SOUND_FILE)
    if len(audio_original.shape) > 1:
        audio_original = audio_original[:, 0]
    print(f"Áudio carregado! Shape: {audio_original.shape}, fs: {fs_original} Hz")
else:
    print("Arquivo não encontrado. Gerando áudio de teste...")
    fs_original = 22050
    t_audio = np.linspace(0, 1.5, int(fs_original * 1.5), endpoint=False)
    audio_original = (np.sin(2 * np.pi * 261.63 * t_audio) +
                      np.sin(2 * np.pi * 329.63 * t_audio) +
                      np.sin(2 * np.pi * 392.00 * t_audio))

audio_original = audio_original / np.max(np.abs(audio_original))
N = len(audio_original)

# Análise espectral
espectro = np.abs(rfft(audio_original))
frequencias = rfftfreq(N, 1 / fs_original)

plt.figure(figsize=(10, 4))
plt.plot(frequencias, espectro, 'b')
plt.title("Espectro de Frequências do Áudio")
plt.xlabel("Frequência (Hz)")
plt.ylabel("Magnitude")
plt.grid(True)
plt.show()


energia_total = np.sum(espectro**2)
energia_acumulada = np.cumsum(espectro**2)
indice_energia = np.where(energia_acumulada >= 0.99 * energia_total)[0][0]
f_max = frequencias[indice_energia]

fs_minima = 2 * f_max
print(f"\nFrequência máxima relevante (F_max) ~ {f_max:.2f} Hz")
print(f"Frequência de Nyquist mínima: {fs_minima:.2f} Hz")


fs_maior = int(fs_original * 3)
num_amostras_maior = int(N * (fs_maior / fs_original))
audio_maior = resample(audio_original, num_amostras_maior)


passo = 10
fs_menor = fs_original // passo
audio_menor = audio_original[::passo]

print(f"\nComparação de frequências de amostragem:")
print(f"Original: fs = {fs_original} Hz")
print(f"Maior: fs = {fs_maior} Hz")
print(f"Menor: fs = {fs_menor} Hz")

print("\n PLAYERS DE ÁUDIO ")
print("1. Áudio Original")
display(Audio(audio_original, rate=fs_original))

print(f"\n2. Áudio Superamostrado (fs = {fs_maior} Hz)")
display(Audio(audio_maior, rate=fs_maior))

print(f"\n3. Áudio Subamostrado (fs = {fs_menor} Hz) - Aliasing!")
display(Audio(audio_menor, rate=fs_menor))
```
