## Sinal de voltagem (circuito RC)

<img src="./figures/circuitoRC-01.png" alt="" width="500"/>

<img src="./figures/circuitoRC-02.png" alt="" width="500"/>

```python
# Resposta
import numpy as np
import matplotlib.pyplot as plt

# Parâmetros
V0 = 8.0          # Voltagem aplicada (V)
R = 100.0        # Resistência (Ω)
C = 50e-6          
ctempo = R * C       # Constante de tempo (s)

# Sinal
t = np.linspace(0, 100, 1000)
V_capacitor = V0 * (1 - np.exp(-t / ctempo))

plt.plot(t, V_capacitor)
plt.xlabel('Tempo (s)')
plt.ylabel('Voltagem (V)')
plt.show()
```

## Sinal de balística

<img src="./figures/balistica.png" alt="" width="500"/>

```python
#Função matemática = h(t) = h₀ + v₀×t - ½×g×t²

# Parâmetros
h0 = 1750.0        # Altura inicial (m)
v0 = 10.0         # Velocidade inicial (m/s)
g = 10          # Gravidade (m/s²)

# Sinal
t_max = (v0 + np.sqrt(v0**2 + 2*g*h0)) / g
t = np.linspace(0, 20, 500)
h = h0 + v0*t - 0.5*g*t**2

plt.plot(t, h)
plt.xlabel('Tempo (s)')
plt.ylabel('Altura (m)')
plt.show()
```
